```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time
import copy
```

```python
def Plotter(e):

  fig = plt.figure(figsize=(10,5))
  ax = fig.add_subplot(121)
  ax.set_title('epoch: {:.0f}'.format(e), fontsize=10)

  return fig, ax
```

```python
# Definimos una funcion no convexa
def f(x):
  return x**2 + 10*np.sin(x)
```

```python
xvalues = np.linspace(-10,10,100)
yvalues = f(xvalues)
```

```python
# Definimos la derivada
def Derivative(x,f,h=1e-4):
  return (f(x+h) - f(x-h))/(2*h)
```

```python
def GradientDescent(f, epochs=200, lr=0.05):

  # Necesitamos una semilla para el metodos

  xn = 10.

  for e in range(epochs):
    clear_output(wait=True)

    # El algoritmo
    xn1 = xn - lr*Derivative(xn,f)
    # No se esta actualizando
    xn = xn1

    fig, ax = Plotter(e)
    ax.plot(xvalues,yvalues)
    ax.scatter(xn,f(xn),color='r')

    time.sleep(0.5)

    plt.show()
```

```python
GradientDescent(f)
```

```python
def f(x, y):
  return x**2 + y**2 + 2*x*y
```

```python
def derivadax(x, y, f, h=1e-4):
  return (f(x + h, y) - f(x - h, y))/(2*h)
```

```python
def derivaday(x, y, f, h=1e-4):
  return (f(x, y + h) - f(x, y - h))/(2*h)
```

```python
xvalues = np.linspace(-10,10,100)
yvalues = np.linspace(-10,10,100)
zvalues = f(xvalues,yvalues)
```

```python
def GradientDescent(f, epochs=200, lr=0.05):

  # Necesitamos una semilla para el metodos

  xn = 10.
  yn = 10.

  for e in range(epochs):
    clear_output(wait=True)

    # El algoritmo
    xn1 = xn - lr*derivadax(xn, yn,f)
    yn1 = yn - lr*derivaday(xn, yn,f)
    # No se esta actualizando
    xn = xn1
    yn = yn1

    fig, ax = Plotter(e)
    ax.plot(xvalues,yvalues,zvalues)
    ax.scatter(xn, yn, f(xn, yn), color="black", s=60)

    time.sleep(0.5)

    plt.show()
```

```python
GradientDescent(f)
```
