```python
def Derivative(f,x,h=1e-6):
  return (f(x+h)-f(x-h))/(2*h)
```

```python
def GetNewtonMethod(f,df,xn,itmax=1000,precision=1e-13):

  error = 1
  it = 0

  while error > precision and it < itmax:

    try:
      xn1 = xn - f(xn)/df(f,xn)
      error = np.abs( xn1 - xn )
    except ZeroDivisionError:
      print('Division por zero')
      print(xn)
      return False

    xn = xn1
    it += 1

  if it == itmax:
    return False
  else:
    print(it)
    return xn
```

```python
GetNewtonMethod(Function,Derivative,10.) # la semilla del algoritmo
```

```python
GetNewtonMethod(lambda x : x**2 + 2,Derivative,10.)
```

```python
def funcion(x):
  return 3*x + 1
```

```python
def error(x):
  p_x = 4
  p_y = 5
  f = funcion
  return ((x - p_x)**2 + (f(x) - p_y)**2)**(1/2)
```

```python
GetNewtonMethod(error,Derivative,2.)
```

```python
def Derivative2(f, x, h=1e-6):
    return (f(x+h) - 2*f(x) + f(x-h)) / h**2
```

```python
GetNewtonMethod(
    lambda x: Derivative(error, x),
    lambda f, xn: Derivative2(error, xn),
    2.
)
```
