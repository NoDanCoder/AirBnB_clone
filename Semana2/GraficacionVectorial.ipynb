```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1,1,100)
y = np.linspace(-1,1,100)
```

```python
x
```

```python
y
```

```python
# Generamos un lattice
X,Y = np.meshgrid(x,y)
```

```python
plt.scatter(X,Y)
```

```python
# Definamos la imagen del soporte
def f2(x,y):
  return x**2 + y**2

def f3(x,y):
  return x**2 - y**2
```

```python
# NEcesitamos recorrer la grid para hacer el calculo de la funcion
Z = np.zeros_like(X) # El mapa escalar Z hereda la dim de X
Z
```

```python
Z1 = np.zeros_like(X)
Z1
```

```python
for i in range(x.shape[0]):
  for j in range(y.shape[0]):
    Z[i,j] = f2(x[i],y[j]) # COmo en papel
    Z1[i,j] = f3(x[i],y[j])
```

```python
Z
```

```python
# Queremos grafica
fig = plt.figure()
ax = fig.add_subplot(211,projection='3d')
ax1 = fig.add_subplot(212,projection='3d')

ax.plot_surface(X,Y,Z,cmap='jet')
ax1.plot_surface(X,Y,Z1,cmap='viridis')
```
