```python
def TriangularSuperior(A,b=None):

  A = np.array(A, dtype=float, copy=True)
  n = A.shape[0]

  y = None if b is None else np.array(b, dtype=float, copy =True)

  print(y)

  # VOlver triangular superior
  # Iteramos sobre las filas
  for k in range(n):
    print(k)
    if A[k,k] == 0:
      raise ValueError("NO creemos que gemini sepa todo: pueden haber problemas con la matriz")
    else:
      pivote = A[k,k]
      # Iteramos sobre las columnas
      for i in range(k+1,n):
        print(k,i)
        m_i_k = A[i,k]/pivote
        print('El factor es:', m_i_k)
        # restamos a la fila j, el mik por la fila i
        print('antes',A[i,:])
        A[i, k:] = A[i, k:] - m_i_k*A[k, k:]
        print('despues',A[i,:])
        if b is not None:
          y[i] = y[i] - m_i_k*y[k]

  return (A,y) if b is not None else A
```
